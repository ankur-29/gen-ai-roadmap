# Natural Language Processing (NLP)

## What is NLP?

**Natural Language Processing (NLP)** is a branch of Artificial Intelligence (AI) that enables computers to work with human language.

NLP combines:
- Linguistics
- Machine Learning
- Deep Learning

### Examples
- Chatbots
- Language translation
- Sentiment analysis
- Spam detection
- Text summarization
- Voice assistants
- Question answering systems


## NLP Pipeline

A typical NLP system can contain the following stages:

1. Text collection
2. Text cleaning
3. Sentence tokenization
4. Word tokenization
5. Stop-word handling
6. Stemming / Lemmatization
7. Part-of-Speech tagging
8. Named Entity Recognition
9. Feature extraction
10. Model training
11. Prediction / response generation


## Simple Example

Input:

> I love this movie!

Possible NLP tasks:

- **Sentiment:** Positive
- **Emotion:** Happiness
- **Intent:** Expressing satisfaction

The same sentence can therefore be analyzed in different ways depending on the NLP problem.


## Applications of NLP

| Application | Example |
|---|---|
| Sentiment Analysis | Positive/negative reviews |
| Machine Translation | English → Hindi |
| Chatbots | Customer support |
| Spam Detection | Email classification |
| Text Summarization | News summaries |
| Question Answering | Answering user questions |
| Speech Recognition | Voice → text |


## Limitations of NLP

NLP systems can struggle with:

- Sarcasm
- Ambiguous words
- Context
- Slang
- Multiple meanings
- Low-resource languages
- Very large or domain-specific vocabularies
