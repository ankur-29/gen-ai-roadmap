# NLP Text Preprocessing

Text preprocessing converts raw text into a form that machine-learning models can work with more easily.


```python
import re

text = "I LOVE NLP! NLP is amazing. Visit https://example.com"

print(text)
```

## 1. Lowercasing

```python
text_lower = text.lower()
print(text_lower)
```

## 2. Remove URLs and punctuation

```python
clean_text = re.sub(r'https?://\\S+|www\\.\\S+', '', text_lower)
clean_text = re.sub(r'[^a-zA-Z0-9\\s]', '', clean_text)

print(clean_text)
```

## 3. Tokenization

```python
tokens = clean_text.split()
print(tokens)
```

## 4. Stop Words

Stop words are common words that may provide limited information for some NLP tasks.

Examples:

`the`, `is`, `a`, `an`, `and`, `of`

However, stop-word removal is **task dependent**. For sentiment analysis, words such as `not` can be important.


## 5. Stemming vs Lemmatization

```python
# Example idea:
words = ["playing", "played", "plays", "player"]

# Stemming attempts to reduce words to a root-like form.
# Lemmatization attempts to return a meaningful dictionary form.

print(words)
```
