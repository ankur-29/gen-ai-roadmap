# NLP Feature Extraction

Machine-learning algorithms generally need numerical input. NLP therefore converts text into numerical representations.

Important techniques:

1. Bag of Words
2. TF-IDF
3. Word embeddings
4. Transformer-based representations


## 1. Bag of Words

Bag of Words represents a document using word counts.

Example:

- Document 1: `I love NLP`
- Document 2: `I love Python`

Vocabulary:

`I, love, NLP, Python`

The documents can then be represented numerically using word frequencies.


```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "I love NLP",
    "I love Python",
    "NLP is useful"
]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(documents)

print("Vocabulary:", vectorizer.get_feature_names_out())
print(X.toarray())
```

## 2. TF-IDF

**TF-IDF (Term Frequency-Inverse Document Frequency)** gives higher importance to words that are frequent in a document but less common across the entire collection.

It is commonly used for:
- Text classification
- Search
- Document similarity


```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(documents)

print("Vocabulary:", vectorizer.get_feature_names_out())
print(X.toarray())
```
