# Common NLP Tasks

## 1. Sentiment Analysis

Determine whether text expresses a positive, negative, or neutral sentiment.

Example:

`"The product is excellent"` → Positive


## 2. Text Classification

Assign a category to a document.

Example:

`"Win a free prize now!"` → Spam


## 3. Named Entity Recognition (NER)

Identify entities such as:

- Person
- Organization
- Location
- Date
- Money

Example:

`"Elon Musk founded SpaceX."`

Possible entities:
- Elon Musk → Person
- SpaceX → Organization


## 4. Part-of-Speech Tagging

Identify the grammatical role of words.

Example:

`Anand writes Python code.`

- Anand → Noun
- writes → Verb
- Python → Proper noun
- code → Noun


## 5. Machine Translation

Convert text from one language to another.

Example:

`Good morning` → `सुप्रभात`


## 6. Text Summarization

Generate a shorter version of a longer document while preserving important information.
