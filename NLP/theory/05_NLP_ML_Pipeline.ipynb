# End-to-End NLP Machine Learning Pipeline

A typical classical NLP classification project can follow this workflow:

**Raw Text → Preprocessing → Feature Extraction → Train/Test Split → Model → Evaluation**


```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report

texts = [
    "I love this product",
    "This is an excellent service",
    "I hate this product",
    "This service is terrible",
    "Amazing experience",
    "Very bad experience"
]

labels = [1, 1, 0, 0, 1, 0]

X_train, X_test, y_train, y_test = train_test_split(
    texts,
    labels,
    test_size=0.33,
    random_state=42,
    stratify=labels
)
```

```python
model = Pipeline([
    ("tfidf", TfidfVectorizer()),
    ("classifier", LogisticRegression())
])

model.fit(X_train, y_train)

predictions = model.predict(X_test)

print(classification_report(y_test, predictions))
```

## Predict New Text

```python
new_text = ["I really love this service"]
prediction = model.predict(new_text)

print("Prediction:", prediction[0])
```

## Important Evaluation Metrics

For classification problems, commonly used metrics include:

- Accuracy
- Precision
- Recall
- F1-score
- Confusion matrix

For imbalanced datasets, do not rely only on accuracy.
