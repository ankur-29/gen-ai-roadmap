# Machine Learning (ML) — Introduction

Machine Learning is a branch of Artificial Intelligence in which computers learn patterns from data and use those patterns to make predictions or decisions.

## Main Types of Machine Learning

1. **Supervised Learning** — learns from labeled data.
2. **Unsupervised Learning** — finds patterns in unlabeled data.
3. **Semi-Supervised Learning** — uses a mixture of labeled and unlabeled data.
4. **Reinforcement Learning** — learns through rewards and penalties.

## Examples

- House price prediction → Regression
- Spam detection → Classification
- Customer segmentation → Clustering
- Recommendation systems → Machine Learning


## Typical ML Workflow

**Problem Definition → Data Collection → Data Understanding → Data Cleaning → EDA → Feature Engineering → Train/Test Split → Model Training → Evaluation → Hyperparameter Tuning → Deployment → Monitoring**


## Important ML Terminology

| Term | Meaning |
|---|---|
| Feature | Input variable |
| Target | Variable to predict |
| Sample | One observation/row |
| Training Data | Data used to learn |
| Test Data | Data used to evaluate |
| Model | Learned mathematical representation |
| Parameter | Value learned during training |
| Hyperparameter | Configuration selected before/during training |
