# Data Preprocessing

Data preprocessing prepares raw data for machine learning.

Common steps:

- Handling missing values
- Removing duplicates
- Detecting outliers
- Encoding categorical features
- Feature scaling
- Data type correction


```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Age": [25, 30, np.nan, 40, 35],
    "Salary": [30000, 50000, 45000, np.nan, 70000],
    "City": ["Bangalore", "Delhi", "Bangalore", "Mumbai", "Delhi"]
})

df
```

## Check Missing Values

```python
print(df.isnull().sum())
```

## Numerical Imputation

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())

df
```

## Encoding Categorical Variables

Common encoders:

- `OneHotEncoder` — nominal categories
- `OrdinalEncoder` — ordered categories
- `LabelEncoder` — usually used for target labels


```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
encoded = encoder.fit_transform(df[["City"]])

print(encoder.get_feature_names_out(["City"]))
print(encoded)
```

## Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_salary = scaler.fit_transform(df[["Salary"]])

print(scaled_salary)
```
