# Exploratory Data Analysis (EDA) and Feature Engineering

EDA helps us understand the structure, distributions, relationships, and problems in a dataset.

## EDA Questions

- How many rows and columns?
- What are the data types?
- Are there missing values?
- Are there duplicates?
- What are the distributions?
- Are there outliers?
- Which features are correlated?
- Is the target balanced?


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Age": [22, 25, 30, 35, 40, 45, 50],
    "Salary": [25000, 30000, 42000, 50000, 65000, 75000, 90000]
})

print(df.shape)
print(df.info())
print(df.describe())
```

## Distribution

```python
df["Salary"].hist()
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.title("Salary Distribution")
plt.show()
```

## Correlation

```python
print(df.corr(numeric_only=True))
```

## Feature Engineering

Feature engineering creates or transforms features to improve model performance.

Examples:

- Extract year/month/day from dates
- Create BMI from height and weight
- Log-transform skewed variables
- Create interaction features
- Bin continuous variables
- Aggregate historical values
