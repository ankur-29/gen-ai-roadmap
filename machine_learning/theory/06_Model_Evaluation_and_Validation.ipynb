# Model Evaluation and Validation

A model should be evaluated on data that was not used to train it.

## Common Classification Metrics

- Accuracy
- Precision
- Recall
- F1-score
- ROC-AUC
- Confusion Matrix

## Common Regression Metrics

- MAE
- MSE
- RMSE
- R²
- MAPE


## Cross-Validation

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier

X, y = load_iris(return_X_y=True)

model = RandomForestClassifier(random_state=42)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

scores = cross_val_score(model, X, y, cv=cv, scoring="accuracy")

print("Scores:", scores)
print("Mean CV score:", scores.mean())
```

## Overfitting vs Underfitting

### Overfitting
The model learns the training data too closely and performs poorly on unseen data.

### Underfitting
The model is too simple to capture important patterns.

The goal is to find a model that generalizes well.
