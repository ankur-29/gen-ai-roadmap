```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("student_dataset.csv") # Data gathering
```

```python
data = df['Gender'].value_counts()
data
x = data.index
y = data.values

plt.figure(figsize=(10, 4))
plt.title('Gender Analysis')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.bar(x,y)
#plt.grid()
```

```python
## Dietary Habits vs Depression

df = df[df['Dietary Habits'] != 'Others']
df['Dietary Habits'].value_counts()

data = df.groupby('Dietary Habits')['Depression'].sum()

x = data.index
y = data.values

plt.figure(figsize=(10, 4))
plt.title('Depression Vs Health')
plt.xlabel('Dietary Habits')
plt.ylabel('Count')
plt.bar(x,y)
```

```python
# Gender Vs Academic Pressure

df_male = df[df['Gender'] == 'Male'].head(12348)
df_female = df[df['Gender'] == 'Female']

### join above dataset in vertical way
### concat() --> used for joining multiple datasets

df = pd.concat([df_male, df_female])
data = df.groupby('Gender')['Academic Pressure'].sum()

x = data.index
y = data.values

plt.figure(figsize=(10, 4))
plt.title('Gender based Academic Pressure')
plt.xlabel('Gender')
plt.ylabel('Total Academic Pressure')
plt.bar(x,y, color='g', width=0.2)
plt.show()
```

##### Horizontal bar charts ---> Bars are going to horizontal

```python
### barh() --> used for showing horizontal bar graphs

df_sleep = df[df['Sleep Duration'] != 'Others']
data = df_sleep['Sleep Duration'].value_counts()

plt.figure(figsize=(10, 4))
plt.title('Sleep duration of students')
plt.ylabel('Sleep duration')
plt.xlabel('Count')
plt.barh(data.index, data.values, color='r')
plt.show()
```

##### Multiple bar charts

```python
data = df.groupby("Gender")[["Academic Pressure", "Financial Stress"]].mean()

x = np.arange(len(data.index))
width = 0.35

plt.figure(figsize=(8,5))
plt.bar(x - width/2, data["Academic Pressure"], width, label="Academic Pressure", color='r')
plt.bar(x + width/2, data["Financial Stress"], width, label="Financial Stress", color='g')

plt.xticks(x, data.index) # is used to set or customize the labels and positions on the x-axis.

plt.title("Academic Pressure vs Financial Stress")
plt.xlabel("Gender")
plt.ylabel("Average Score")

plt.legend() # A legend tells us which color or bar represents which data.

plt.show()
```

```python
## Stacked bar chart
### Instead of keeping the bars side by side, we can keep the bars in the stacked way i.e. one above another

# Plot first bar
plt.bar( data.index, data["Academic Pressure"], label="Academic Pressure")

# Stack second bar on top of first
plt.bar(data.index, data["Financial Stress"], bottom=data["Academic Pressure"],label="Financial Stress")

plt.title("Academic Pressure and Financial Stress by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Score")
plt.legend()
plt.show()
```
