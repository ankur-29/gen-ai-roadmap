```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("student_dataset.csv") # Data gathering

df["Sleep Duration"].value_counts()
```

### Univariate analysis and Bivariate analysis

#### Univariate analysis --> Analysis or will be creating visualization for given unique column
##### Example : How value are spread in age column

#### Bivariate analysis --> Analysis of 2 columns
##### Example : Month on Month sales of product for company

#### Multivariate analysis --> more than 2 columns

```python
# 2D line plot
## specific purpose --> to show the relationship between two variables
## usually work with categorical-categorical, categorical-numerical, numerical-numerical
#### Y axis --> dependent variable or value that are going to change for that category
#### X axis --> independent variable or Category variable

data = df["Sleep Duration"].value_counts()
x = data.index
y = data.values
plt.figure(figsize=(10,4))
plt.title('Count of population as per Sleep Duration', color='red')
plt.ylabel('Poputaltion count')
plt.xlabel('Sleep Duration', color='green')
plt.plot(x, y, color="#23ccff", marker='o', linestyle='--', linewidth=1.5, markersize=8)
plt.grid()
```
