```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("student_dataset.csv") # Data gathering
```

```python
df['Age'] = df['Age'].astype('int32')

df.info()
```

```python
df['CGPA'] = df['CGPA'].astype('float32')

df['Gender'] = df['Gender'].astype('category')
df['City'] = df['City'].astype('category')
df['Profession'] = df['Profession'].astype('category')
df['Sleep Duration'] = df['Sleep Duration'].astype('category')
df['Dietary Habits'] = df['Dietary Habits'].astype('category')
df['Degree'] = df['Degree'].astype('category')

df.info()
```

```python
## line plot --> AGE vs CGPA

data = df.groupby('Age')['CGPA'].mean().head(10)
x = data.index
y = data.values

plt.figure(figsize=(10, 4))
plt.title("CGPA distribution as per Age", color='r')
plt.ylabel('CGPA')
plt.xlabel('AGE')
plt.plot(x, y, color="#ff23bd", marker='o', linestyle='--', linewidth=1.5, markersize=8)
plt.grid()
```

```python
## line plot --> Provide the mean stress value of top 10 populated cities.

x = df['City'].value_counts().head(10)
y = df.groupby('City')['Financial Stress'].mean()
y = y[(y.index == 'Kalyan') | (y.index == 'Srinagar') | (y.index == 'Hyderabad') | (y.index == 'Vasai-Virar') 
      | (y.index == 'Lucknow') | (y.index == 'Thane') | (y.index == 'Ludhiana') | (y.index == 'Agra') 
      | (y.index == 'Surat') | (y.index == 'Kolkata')]

x = x.sort_index().index
y = y.values

plt.figure(figsize=(10, 4))
plt.title("Top 10 Cities Mean Financial Stress")
plt.ylabel('City')
plt.xlabel('Mean Financial Stress')
plt.plot(x, y, color="r", marker='o', linestyle='--', linewidth=1.5, markersize=8)
plt.grid()
```
