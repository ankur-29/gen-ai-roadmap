```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("student_dataset.csv") # Data gathering
```

#### Pie chart --> used for Bivariate analysis and is being used for relating Categorical-numberical data.

```python
# depression distribution
depression_counts = df["Depression"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(depression_counts, labels=["No", "Yes"],autopct="%1.1f%%")
plt.title("Depression Distribution")
plt.show()
```

```python
# health distribution
df = df[df['Dietary Habits'] != 'Others']
data = df['Dietary Habits'].value_counts()


plt.figure(figsize=(6,6))
plt.pie( data, 
    labels=data.index, 
    autopct="%1.1f%%", 
    colors=['red', 'green', 'pink'],
    explode=[0.1, 0.0, 0.0]
)
plt.title("Health Distribution")
plt.legend()
plt.show()
```
