```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("student_dataset.csv") # Data gathering
```

##### Scatter plot --> show the relationship between two numerical variables
##### It figure outs co-relation between two numerical values

```python
### Age VS average CGPA
data = df.groupby('Age')['CGPA'].mean().head(10)
x = data.index
y = data.values

plt.figure(figsize=(10, 4))
plt.title('Average CGPA variation as per AGE')
plt.xlabel('Age')
plt.ylabel('Average CGPA')
plt.scatter(x,y, color="#ff34dd", marker='*')
plt.grid()
```
