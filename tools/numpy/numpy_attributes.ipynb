# Attributes

```python
import numpy as np
```

```python
# ndir :--> Get the number of dimensions of the array
arr = np.array([1, 2,3])
arr.ndim  #1

arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
arr_2d.ndim  #2
```

```python
#shape :--> provide the idea about rows and columns of the array/dataset

arr.shape  #(3,)
arr_2d.shape  #(2, 3)
```

```python
#size :--> Get the total number of elements in the array

arr.size  #3
arr_2d.size  #6
```

```python
#itemsize :--> Get the size in bytes of each element in the array

arr.itemsize  #4 (for int32)
arr_2d.itemsize  #4 (for int32)
```
