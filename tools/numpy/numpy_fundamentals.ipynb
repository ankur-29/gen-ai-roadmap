# Numpy

#### Numpy is a fundamental Package for scientific computing in python. It is a Pthon library that provides a multi dimensional Array object and various derived objects such as Masked Arrays, Matrices and an assortment of routines for fast operations on Array include Mathematical, logical, Shape manipulation, Sorting, Selecting, I/O, Descrete Fourier Transform, basic Linear Algebra, Stastical Operations, Random simulation, etc

## Numpy Arrays Vs Python Sequences

#### Numpy Array have a fixed size at creation, unlike Python lists which row dynamically.Changing the size of an n-dim array will create a new array and delete the original.

### The element in Numpy array are req to have same datatype and thus will be of same size in memory.

### Numpy array facilitates advanced Mathematical and other types of operation on large number of data. typically, such operation are executed more efficently with less code than is possible using python's built in sequences.

## dtype parameter ---> help us out with deciding about size of data we are going to store inside our array

```python
np.array([1, 2, 3, 4, 5], dtype=np.int32) # memory size of 4 bytes per element

np.array([11, 21, 32, 4, 53], dtype=np.int64) # memory size of 8 bytes per element
```

## arange() ---> It helps us out with providing the values over a certain range

```python
np.arange(10) # 1D array with values from 0 to 9
```

```python
np.arange(5, 25) # 1D array with values from 5 to 24

np.arange(5, 25, 2) # 1D array with values from 5 to 24 with step size of 2
```

```python
arange_data = np.arange(1, 17)

arange_data.reshape(4, 4) # Reshape the 1D array into a 2D array with 4 rows and 4 columns
```

```python
np.ones((3, 3)) # Create a 2D array of shape (3, 3) filled with ones
```

```python
np.zeros((2, 4)) # Create a 2D array of shape (2, 4) filled with zeros
```

```python
np.random.random((3, 3)) # Create a 2D array of shape (3, 3) filled with random values between 0 and 1
```

```python
np.identity(4) # Create a 2D identity matrix of shape (4, 4)
```

```python
np.linspace(0, 1, 5) # Create a 1D array of 5 evenly spaced values between 0 and 1
```
