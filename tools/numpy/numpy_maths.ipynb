```python
import numpy as np
```

# Mathematical operation in numpy

```python
arr = np.random.random((3, 3))* 100
```

```python
arr = np.round(arr)

arr
```

```python
# Basic Arithmetic Functions

# addition
arr = np.add(arr, 10)
print(arr)
print("-" * 40)

# subtraction
arr = np.subtract(arr, 10)
print(arr)
print("-" * 40)

# multiplication
arr = np.multiply(arr, 10)
print(arr)
print("-" * 40)

# division
arr = np.divide(arr, 10)
print(arr)
print("-" * 40)
```

```python
# vector operations
arr_1 = np.arange(1, 17).reshape(4, 4)
arr_2 = np.arange(17, 33).reshape(4, 4)

print(arr_1 + arr_2)
print("-" * 40)

print(arr_2 - arr_1)
print("-" * 40)
```

```python
# axis in numpy is used to specify the dimension along which an operation should be performed.
# axis=0 --> rows (vertical direction), and axis=1 --> columns (horizontal direction).

array_data = np.array(np.random.randint(30, 77, size=(4,4)))
print(array_data)
print("-" * 40)

print(np.max(array_data))
print("Max along axis 0 (rows):" , np.max(array_data, axis=0))
print("Max along axis 1 (columns):" , np.max(array_data, axis=1))
```

```python
print(np.min(array_data))
print("Min along axis 0 (rows):" , np.min(array_data, axis=0))
print("Min along axis 1 (columns):" , np.min(array_data, axis=1))
print("-" * 40)

print(np.sum(array_data))

```
