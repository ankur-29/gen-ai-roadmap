```python
import numpy as np

array_data = np.array(np.random.randint(30, 77, size=(4,4)))
print(array_data)
```

# Statistical Functions

```python
# mean
print("Mean of the array:", np.mean(array_data))
print("-" * 40)

# median
print("Median of the array:", np.median(array_data))
print("-" * 40)

# standard deviation
print("Standard Deviation of the array:", np.std(array_data))
print("-" * 40)

# variance
print("Variance of the array:", np.var(array_data))
print("-" * 40)
```

# log and exponential functions

```python
arr = np.array([1, 2, 3])

# natural logarithm
print("Natural Logarithm of the array:", np.log(arr))
print("-" * 40)

# Base 10 logarithm
print("Base 10 Logarithm of the array:", np.log10(arr))
print("-" * 40)

#Base-2 logarithm
print("Base 2 Logarithm of the array:", np.log2(arr))
print("-" * 40)

#exponential
print("Exponential of the array:", np.exp(arr))
print("-" * 40)
```
