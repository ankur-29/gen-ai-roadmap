```python
import pandas as pd
import numpy as np
```

```python
# DataFrame --> Combination of multiple Series objects, where each Series represents a column in the DataFrame. 
# Each column can have a different data type (e.g., integers, floats, strings, etc.).
# The DataFrame is a two-dimensional labeled data structure with rows and columns.

# read_csv() --> function is used to read the CSV file and create a DataFrame
df = pd.read_csv("youtuber_names.csv")

df2 = pd.read_csv("youtube_first_two_columns.csv")

df.head() # Display the first 5 rows of the DataFrame
```

```python
# Display the rows, columns of dataframe
print(df.shape)
print(df2.shape)
print("-" * 40)

# Display the total number of elements in the DataFrame
print(df.size)
print(df2.size)
print("-" * 40)

# ndim --> attribute is used to get the number of dimensions of the DataFrame
print(df.ndim)
print(df2.ndim)
print("-" * 40)

# columns --> attribute is used to get the column labels of the DataFrame
print(df.columns)
print(df2.columns)
print("-" * 40)
```
