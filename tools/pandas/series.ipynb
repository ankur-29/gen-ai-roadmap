```python
import pandas as pd
import numpy as np

print(pd.__version__)
print(np.__version__)
```

## Pandas Series is column of data in a Table like in Excel or Google sheets with labels called as indexes attached to each value

```python
pd.Series(data=[40, 50, 60], index=['Bread', 'Butter', 'Milk'], name='Grocery Items')
```

## Series is a 1D Array that can hold an type of data like integers, float, strings, boolean

```python
# Creating series in different ways

# 1. From a list
country_list = ['India', 'USA', 'UK', 'Germany', 'France']
s1 = pd.Series(country_list)
print(s1)
print("-" * 40)

# 2. From a numpy array
subject_names = np.array(['Math', 'Science', 'English', 'History', 'Geography'])
subject_marks = np.array([71, 61, 79, 89, 82])
s2 = pd.Series(subject_marks, index=subject_names, name='Subject Marks')
print(s2)
print("-" * 40)

# 3. From a dictionary
data_items = {"maths": 90, "science": 80, "english": 70, "history": 60}
s3 = pd.Series(data_items, name='Subject Marks')
print(s3)
print("-" * 40)

# 4. Custom index to series
values = [10, 20, 30, 40, 50]
custom_index = ['a', 'b', 'c', 'd', 'e']
s4 = pd.Series(values, index=custom_index, name='Custom Index Series')
print(s4)
print("-" * 40)

# Scalar value to series
scalar_value = 100
s5 = pd.Series(scalar_value, index=['x', 'y', 'z'], name='Scalar Value Series')
print(s5)
```

## Series Methods

```python
df = pd.read_csv("youtuber_names.csv") # dataframe
series_from_df = df['Youtuber Name'] # series

# head(n) --> returns the first n rows of the series, default n=5
print(series_from_df.head(10))

# tail(n) --> returns the last n rows of the series, default n=5
print(series_from_df.tail(10))

# value_counts() --> returns the count of unique values in the series
print(series_from_df.value_counts())

# sort_values() --> returns the series sorted in ascending order
print(series_from_df.sort_values())

# sort_values(ascending=False) --> returns the series sorted in descending order
print(series_from_df.sort_values(ascending=False))

# unique() --> returns the unique values in the series
print(series_from_df.unique())
```

```python
# method chaining --> use of multiple methods in a single line
series_from_df.head().value_counts()

# sort_index() --> returns the series sorted by index
print(series_from_df.sort_index())
```
