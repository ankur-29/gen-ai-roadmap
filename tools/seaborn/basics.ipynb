### Seaborn is a data visualization lib built on top of Matplotlib and is integrated with Python.

##### Sntax --> Seaborn is simpler, less code req compare to Matplotlib
##### More varient of visualization than matplotlib
##### Matplotlib integrates with all DataTypes where as Seaborn integrates with Pandas Dataframe


### There are 6 class plot in Seaborn
##### a. Relational ---> ScatterPlot, LinePlot
##### b. Distribution ---> Histrogram, KDE, RugPlot
##### c. Categorical ---> BarPlot, countPlot, BoxPlot, Violin, swarmPlot
##### d. Matrix ---> HeatMap, clusterMap
##### e. Regression plot ---> regPlot
##### f. Multiplot ---> joinPlot, pairPlot
