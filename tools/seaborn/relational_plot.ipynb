```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df = pd.read_csv("student_dataset.csv") # Data gathering

### lineplot() ---> used for ploting line
sns.lineplot(df, x='Age', y='CGPA')
```

```python
### relplot() ---> capabble of ploting both scatterplot and lineplot based on kind parameter
### kind = 'line' --> lineplot, kind='scatter' --> scatterplot

sns.relplot(data=df, x='Age', y='CGPA', kind='line')
sns.relplot(data=df, x='Age', y='CGPA', kind='scatter')
```
